# A short plot of the nucleon fractions in the early universe

```python
import numpy as np 
import matplotlib.pyplot as plt

T = np.logspace(9,13, 1000) #temperature range from 10 to 10^15 K
kb = 1.380649e-23 # Boltzmann constant in J/K
c = 3e8 # speed of light in m/s
Delta_m = 2.3e-30 # mass difference in kg

p_fraction = 1/(1+np.exp(-Delta_m*c**2/(kb*T))) # proton fraction
n_fraction = np.exp(-Delta_m*c**2/(kb*T)) / (1+np.exp(-Delta_m*c**2/(kb*T))) # neutron fraction

plt.figure(figsize=(10,6))
plt.plot(T, p_fraction, label='Proton Fraction', color='blue')
plt.plot(T, n_fraction, label='Neutron Fraction', color='red')
plt.xscale('log')
plt.xlabel('Temperature (K)')
plt.ylabel('Fraction')
plt.title('Proton and Neutron Fractions vs Temperature')
plt.legend()
plt.grid()
plt.savefig('proton_neutron_fractions.png', dpi=500)
plt.show()  
```
