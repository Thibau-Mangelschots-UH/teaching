```python
import numpy as np
import matplotlib.pyplot as plt

hbar = 1.0545718e-34  # Planck's constant over 2 pi in J*s
kb = 1.380649e-23  # Boltzmann constant in J/K
omega = .5e13  # Frequency in rad/s
T = np.logspace(-2,2,1000)
N = 1e23

#point where kb*T = hbar*omega
T_critical = hbar*omega/(kb)

E = (3*N/2)*hbar*omega*(1/np.tanh(hbar*omega/(2*kb*T)))


plt.plot(T, E/(kb))
plt.xlabel('Temperature (K)')
plt.ylabel(r'$E/k_B$')
plt.axvline(x=T_critical, color='r', linestyle='--', label=r'$k_B T = \hbar \omega$ ')
plt.axhline(y=3*N/2*hbar*omega/(kb), color='g', linestyle='--', label=r'$E = \frac{3}{2}N\hbar\omega$ ')
plt.title('Energy vs Temperature for a Quantum Harmonic Oscillator')
plt.grid()
plt.legend()
plt.savefig('energy_vs_temperature.png')
```

```python
C = np.gradient(E, T)
plt.plot(T, C/kb)
plt.xlabel('Temperature (K)')
plt.ylabel(r'$C_V/k_B$')
plt.axvline(x=T_critical, color='r', linestyle='--', label=r'$k_B T = \hbar \omega$ ')
plt.axhline(y=3*N, color='g', linestyle='--', label=r'$C =3Nk_B$ ')
plt.title('Heat Capacity vs Temperature for a Quantum Harmonic Oscillator')
plt.grid()
plt.legend()
plt.savefig('heat_capacity_vs_temperature.png')
```
